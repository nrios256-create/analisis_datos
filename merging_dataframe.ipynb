```python
import pandas as pd
```

```python
movie = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/main/movies.csv')
ratings = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/main/ratings.csv')
#esto lo traemos de github del profe
```

```python
print(movie.shape)
print(ratings.shape)
```

```python
print('movie dataset')
print(movie.head())
print('')
print(50*'*')
print('rating dataset')
print(ratings.head())
```

```python
movie_ratings = movie.merge(ratings, on='id')
movie_ratings.head()
movie_ratings.drop(['Unnamed: 0_x','Unnamed: 0_y'],axis=1,inplace=True)
#unir la dataframe
```

```python
movie_ratings = movie.merge(ratings, on='id',suffixes=('_movie','_rating'))
movie_ratings.head()
```

```python
movie_genres = pd.read_csv('https://raw.githubusercontent.com/Gustavo-Bruges/movies/refs/heads/main/movie_to_genres.csv')
```

```python
movie_genres.head()
```

```python
movie_genres = movie_genres.drop(columns=['Unnamed: 0'], axis=1, errors='ignore') #axis=1 : avisarle que me avise la columna, #errors='ignore' : si la columna existe la elimina, de lo contrario sigue con el codigo normal
movie_genres.head()
```

```python
print(movie.head())
print(movie_genres.head())
```

```python
movie_generos = movie.merge(movie_genres,left_on='id',right_on='movie_id')
movie_generos.head()
```

```python
crimen = movie_generos[movie_generos['genre']=='Crime']
```

```python
menos_popular = crimen.loc[crimen['popularity'].idxmin()]
menos_popular.head()
#aqui uso loc para buscar las fila
```

```python
sci_movie = movie_generos[movie_generos['genre']== 'Science Fiction']
print(sci_movie.shape)
action_movie = movie_generos[movie_generos['genre']== 'Action']
print(action_movie.shape)
```

```python
print(sci_movie.head())
print(sci_movie.shape)
print(action_movie.head())
print(action_movie.shape)
```

```python
df_left= action_movie.merge(sci_movie, on='movie_id', how='left', suffixes=('_action','_scifi'))
df_left.head()
#demostrar la union izquierda
#aqui se muestra que una pelicula se puede tomar para varios generos
```

```python
df_right= action_movie.merge(sci_movie, on='movie_id', how='right', suffixes=('_action','_scifi'))
df_right.head()
#demostrar la union derecha
```
