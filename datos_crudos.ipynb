```python
import pandas as pd
import numpy as np
#importo librerias a utilizar
```

```python
datos_cr = pd.read_csv("datos_crudos_tarea.csv") #cargo la base de datos
datos_cr.head() #reviso que si sea la base de datos correcta
```

```python
datos_cr.info()
datos_cr.shape
#reviso la informacion de cada columna
```

```python
datos_cr.columns #consulto el nombre de las columnas para tenerlas claras y ver si hay errores para luego corregirlos
```

Revisé la base y pues sí tiene varios datos faltantes en columnas como sexo, glucosa y colesterol. también miré si había duplicados y no hay y vi que hay una columna que se llama "unnamed: 0" que básicamente no sirve para nada, parece que es solo un índice que se guardó ahí.

```python
datos_cr = datos_cr.drop(columns=['Unnamed: 0'])  # uso drop para eliminar esa columna porque no sirve, es solo un índice que se guardó ahí y no aporta nada
```

```python
datos_cr.columns #aqui confirmo que ya no este la columna unnamed:0
```

```python
datos_cr.head(20) #aqui reviso como se ve el dataset sin esa columna
```

```python
datos_cr.columns = datos_cr.columns.str.strip().str.lower().str.replace(' ', '_') # # arreglo los nombres de las columnas quitando espacios, poniendo todo en minúscula y usando guion bajo para que sea más ordenado
#str.strip() = quita espacios al inicio y final
#str.lower() = pone todo en minúsculas
#str.replace(' ', '_') = cambia espacios por guion bajo
datos_cr.columns
```

```python
datos_cr.duplicated().sum() #revisar si hay duplicados
```

```python
datos_cr[datos_cr.duplicated()] #aqui le pido que me muestre solo las filas que estan duplicadas
```

```python
datos_cr[datos_cr.duplicated(keep=False)].sort_values(by='id_paciente')
# aquí le estoy pidiendo que me muestre todas las filas que están duplicadas, no solo una parte,
# por eso uso keep=False para ver tanto la original como la repetida,
# y luego las ordeno por id_paciente con sort_values para poder verlas mejor organizadas
```

```python
datos_cr = datos_cr.drop_duplicates() #con este drop pido que elimine los duplicados
datos_cr.duplicated().sum() #aqui reviso que queden eliminados
```

```python
datos_cr.isnull().sum()  # aqui reviso cuantos valores faltantes hay en cada columna
```

```python
datos_cr['sexo'].unique()  # aqui reviso las diferentes formas en que está escrita la variable sexo
```

```python
datos_cr['sexo'] = datos_cr['sexo'].str.strip().str.upper()
# aqui limpio los textos quitando espacios con strip y poniendo todo en mayúscula con upper

datos_cr['sexo'] = datos_cr['sexo'].replace({
    'MUJER': 'F',
    'F': 'F',
    'HOMBRE': 'M',
    'M': 'M',
    'H': 'M',
    'X': np.nan
})
# aqui unifico las categorias para que solo queden F y M con replace, y convierto valores raros como X en nulos con np.nan como enseño el profe
datos_cr['sexo'].unique() #reviso que si se aplique lo anterior
```

```python
datos_cr['fuma'].unique()  #reviso como están escritos los valores de los fumadores
```

```python
datos_cr['fuma'] = datos_cr['fuma'].str.strip().str.title()
# aqui limpio los textos quitando espacios con strip y poniendo primera en mayuscula y demas en minuscula

datos_cr['fuma'] = datos_cr['fuma'].replace({
    'SI': 'SI',
    'SÍ': 'SI',
    'Fumador': 'SI',
    'NO': 'NO',
    '1': np.nan,
    '0': np.nan
})
# aqui unifico las formas de si y no utilizando replace, y en el caso de 1 y 0 los paso a nulos porque pues el profe no hizo aclaracion al respecto
#de como tratar 1 y 0 y no tiene mucho sentido que pongan valores numericos asi que los dejare como valores nulos por la falta de informacion

datos_cr['fuma'].unique()
# aqui reviso que ya quede limpio y solo tenga SI, NO o nulos(nan)
```

```python
datos_cr['actividad_fisica'].unique()  # aqui reviso como están escritos
```

```python
datos_cr['actividad_fisica'] = datos_cr['actividad_fisica'].str.strip().str.title()
# aqui limpio los textos quitando espacios con strip y poniendo primera en mayuscula y demas en minuscula

datos_cr['actividad_fisica'] = datos_cr['actividad_fisica'].replace({
    'Alta': 'Alta',
    'Media': 'Media',
    'Baja': 'Baja',
    'Sedentario': 'Sedentario',
    '0': np.nan,
    '1': np.nan,
    '2': np.nan
})
# aqui unifico las categorias para que solo queden niveles claros de actividad fisica,
# y los valores 0, 1 y 2 los paso a nulos porque no es claro que representan y pues aqui no es logico poner numeros

#en la variable “sedentario” decidí dejarla como una categoría aparte, porque aunque podría interpretarse como baja actividad,
#no es exactamente lo mismo porque tambien podria ser que no hace nada de ejercicio y preferí no asumir cosas que no están claras en los datos.

datos_cr['actividad_fisica'].unique()
# aqui reviso que ya quede limpio
```

```python
datos_cr.dtypes #ver qué tipo de dato tiene cada columna y asi saber cuales estan mal y necesito convertir
```

revisé los tipos de datos de cada columna y me di cuenta que varias variables numéricas como edad, peso, altura, glucosa y colesterol están como texto (object), por lo que es necesario convertirlas a numéricas. también la fecha está como texto y debería convertirse a formato de fecha.

```python
datos_cr['edad'] = pd.to_numeric(datos_cr['edad'], errors='coerce')
datos_cr['peso_kg'] = pd.to_numeric(datos_cr['peso_kg'], errors='coerce')
datos_cr['altura_m'] = pd.to_numeric(datos_cr['altura_m'], errors='coerce')
datos_cr['glucosa_mg_dl'] = pd.to_numeric(datos_cr['glucosa_mg_dl'], errors='coerce')
datos_cr['colesterol_total'] = pd.to_numeric(datos_cr['colesterol_total'], errors='coerce')
# aqui convierto cada variable de texto que este mal a número con pd.to_numeric,
#y uso errors='coerce' para que los valores raros o mal escritos se vuelvan nulos
```

```python
datos_cr[['edad', 'peso_kg', 'altura_m', 'glucosa_mg_dl', 'colesterol_total']].describe()
#después de convertir las variables a numéricas, revisé sus estadísticas para identificar posibles valores fuera de rango o inconsistentes en los datos
```

al revisar las estadísticas encontré varios valores fuera de rango en diferentes variables. por ejemplo, hay edades negativas y valores muy altos como 999, pesos negativos o exagerados, alturas imposibles y valores negativos en glucosa y colesterol. Esto indica que hay errores en los datos que deben ser tratados.

```python
# EDAD
datos_cr.loc[(datos_cr['edad'] < 0) | (datos_cr['edad'] > 100), 'edad'] = np.nan
# aqui uso loc para ubicar filas especificas que cumplen una condición, en este caso edades negativas o muy altas
# uso los paréntesis con condiciones y el operador | que significa "o" para filtrar esos valores
# luego en esas mismas filas reemplazo la columna edad por np.nan, que significa un valor nulo
# hago esto porque esas edades no tienen sentido en la vida real
#basicamente le digo que solo tome valores en edad mayor a 0 y menor a 100

# PESO
datos_cr.loc[(datos_cr['peso_kg'] < 30) | (datos_cr['peso_kg'] > 440), 'peso_kg'] = np.nan
# aqui hago lo mismo, uso loc para encontrar pesos muy bajos o exagerados

# ALTURA
datos_cr.loc[(datos_cr['altura_m'] < 1) | (datos_cr['altura_m'] > 2.72), 'altura_m'] = np.nan
## aqui filtro alturas que no son posibles usando loc

# GLUCOSA
datos_cr.loc[(datos_cr['glucosa_mg_dl'] < 0), 'glucosa_mg_dl'] = np.nan
# aqui uso loc para encontrar valores negativos en glucosa
# y los paso a np.nan porque no es lógico tener glucosa negativa

# COLESTEROL
datos_cr.loc[(datos_cr['colesterol_total'] < 0), 'colesterol_total'] = np.nan

#para todo lo demas basicamente es lo mismo pero con otras variables
```

```python
datos_cr[['edad', 'peso_kg', 'altura_m', 'glucosa_mg_dl', 'colesterol_total']].describe()
# aqui vuelvo a revisar las estadísticas después de la limpieza para ver si ya se eliminaron los valores fuera de rango
```

```python
datos_cr.isnull().sum()
# aqui reviso si aumentaron los nulos, porque convertí valores inválidos en NaN
```

después de limpiar los valores fuera de rango, volví a revisar las estadísticas y ya los datos quedaron dentro de rangos más lógicos y realistas pues tuve que investigar cositas que no sabia para poder cuadrar los parametros reales. también aumentaron los valores nulos, lo cual tiene sentido porque muchos datos incorrectos se reemplazaron por NaN.

```python
datos_cr['imc_calculado'] = datos_cr['peso_kg'] / (datos_cr['altura_m'] ** 2)

# aqui calculo el imc usando peso y altura, ya que esta columna estaba vacía
```

```python
datos_cr['fecha_consulta'] = pd.to_datetime(datos_cr['fecha_consulta'], errors='coerce')
# aqui convierto la columna fecha_consulta a formato fecha porque estaba como texto con pd.to_datetime
# uso errors='coerce' para que si llega a haber alguna fecha mal escrita se vuelva nula
```

```python
datos_cr['fecha_consulta'].dtype
#reviso si ya quedó en formato fecha
```

```python
datos_cr.head(60)
```

```python
datos_cr[['presion_arterial']].info()
#reviso tipo es presion_arterial
```

```python
datos_cr[['presion_sistolica', 'presion_diastolica']] = datos_cr['presion_arterial'].str.split('/', expand=True)
# aqui separo la presion arterial en dos partes usando el slash como separador
# porque en esa columna vienen dos valores en uno solo
# uso split para separar el texto usando el "/" como referencia,
# y expand=True hace que cada parte se vaya a una columna diferente en lugar de quedar en una lista
```

```python
datos_cr['presion_sistolica'] = pd.to_numeric(datos_cr['presion_sistolica'], errors='coerce')
datos_cr['presion_diastolica'] = pd.to_numeric(datos_cr['presion_diastolica'], errors='coerce')
# aqui convierto cada parte a número con pd.to_numeric
# uso errors='coerce' para que valores raros como "baja" se vuelvan nulos
```

```python
datos_cr['presion_arterial'] = datos_cr['presion_sistolica'].astype(str) + '/' + datos_cr['presion_diastolica'].astype(str)
# aqui vuelvo a unir las dos columnas en una sola con formato numero/numero
```

```python
datos_cr = datos_cr.drop(columns=['presion_arterial'])
# aqui elimino la columna original porque ya separé la información en dos variables más claras con drop
```

```python
datos_cr.head(60)
```

```python
datos_cr.loc[
    (datos_cr['presion_sistolica'].isna()) |
    (datos_cr['presion_diastolica'].isna()) |
    (datos_cr['presion_sistolica'] < 70) |
    (datos_cr['presion_sistolica'] > 250) |
    (datos_cr['presion_diastolica'] < 40) |
    (datos_cr['presion_diastolica'] > 150),
    ['presion_sistolica', 'presion_diastolica']
] = np.nan
# aqui uso loc para buscar las filas donde alguna de las dos presiones esté nula o tenga valores fuera de rango
# con isna reviso si hay valores faltantes
# y con las demás condiciones reviso si los números son absurdos
# si algo está mal, pongo ambas columnas en nulo porque la presión arterial ya no sería confiable completa
#el | lo uso para decir que si se cumple una condición o la otra, entonces se haga algo
```

```python
datos_cr[['presion_sistolica', 'presion_diastolica']].head(60)
# aqui reviso como quedaron las dos columnas después de la limpieza
```

la presión arterial venía en una sola columna con dos valores, por lo que la separé en presión sistólica y diastólica. la columna original ya no era necesaria, así que la eliminé. entonces para los valores faltantes, decidí mantenerlos como nulos cuando alguna de las dos partes no era válida, para no asumir información incorrecta.

```python
datos_cr.loc[(datos_cr['imc_calculado'] < 10) | (datos_cr['imc_calculado'] > 60), 'imc_calculado'] = np.nan
# aqui uso loc para ubicar las filas donde el imc cumple ciertas condiciones
# uso (condicion1) | (condicion2) donde el | significa "o", es decir, valores menores a 10 o mayores a 60
# esos rangos no son realistas para el imc, por eso los considero errores
# y en esas filas reemplazo el valor por np.nan, que representa un valor nulo
# esto lo hago para eliminar datos que no son confiables sin borrar la fila completa
datos_cr[['imc_calculado']].describe()
```

```python
datos_cr['fecha_consulta']= pd.to_datetime(datos_cr['fecha_consulta'], dayfirst=False,infer_datetime_format=True, errors='coerce')
# aqui convierto la columna fecha_consulta a formato fecha (datetime)
# para poder trabajarla correctamente y que no quede como texto
datos_cr[['fecha_consulta']].info()
```

```python
datos_cr.to_csv('datos_limpios.csv', index=False)
# finalmente guardo la base final limpia
#el index=False lo uso para que no se guarde la numeración de las filas como una columna extra en el archivo csv
```

INTERPRETACIÓN

Profe, al final de todo el proceso lo que hice fue limpiar la base paso a paso para que quedara más organizada y sobre todo más confiable. primero eliminé columnas que no aportaban y dejé los nombres de las variables más claros para poder trabajar mejor. después revisé duplicados y los eliminé para asegurarme de que no hubiera registros repetidos.

luego trabajé las variables categóricas como sexo, fuma y actividad_fisica, porque estaban escritas de muchas formas diferentes. ahí lo que hice fue unificar todo para que quedaran solo categorías claras, y los valores que no tenían sentido o no eran claros los dejé como nulos porque no me parecía correcto asumir cosas.

también me di cuenta que varias columnas que deberían ser numéricas estaban como texto, entonces las convertí a números para poder analizarlas bien. después de eso revisé los valores y encontré muchos fuera de rango, como edades negativas, pesos exagerados o alturas imposibles, entonces esos los pasé a nulos porque claramente eran errores en los datos.

en la presión arterial, como venía en una sola columna con dos valores, decidí separarla en presión sistólica y diastólica para poder trabajarla mejor. además, cuando una de las dos partes estaba mal o incompleta, dejé ambas como nulas porque ese dato ya no era confiable completo.

con la fecha, la convertí a formato datetime porque estaba como texto, y también calculé el imc a partir del peso y la altura. después revisé el imc y eliminé valores que no eran realistas para que quedara dentro de un rango lógico.

en cuanto a los valores nulos, decidí no rellenarlos porque esta base es tipo encuesta, entonces no tiene sentido inventar datos como la edad o el peso de una persona que no los reportó.

al final hice una verificación general revisando tipos de datos, nulos, estadísticas y cómo se veía la base, y ya todo quedó más coherente, sin valores raros y listo para análisis. básicamente con todo esto la idea es poder trabajar con datos más confiables y evitar errores en los resultados.
